```python
import os
import shutil

source = "source"
destination = "destination"

# Create the source directory and add some dummy files for demonstration
os.makedirs(source, exist_ok=True)
with open(os.path.join(source, "image1.jpg"), "w") as f:
    f.write("dummy jpg content")
with open(os.path.join(source, "document.txt"), "w") as f:
    f.write("dummy text content")
with open(os.path.join(source, "image2.jpg"), "w") as f:
    f.write("dummy jpg content")

os.makedirs(destination, exist_ok=True)

for file in os.listdir(source):
    if file.endswith(".jpg"):
        shutil.move(
            os.path.join(source, file),
            os.path.join(destination, file)
        )
        print(file, "moved!")

print("All JPG files moved successfully.")
```
